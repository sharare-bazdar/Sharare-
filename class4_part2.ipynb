```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
iris=load_iris()
x=iris.data
y=iris.target
```

```python
df=pd.DataFrame(x, columns=iris.feature_names)
df['target']=y
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivative(x):
  return x*(1-x)
```

```python
input_layer_size=x_train.shape[1]
hidden_layer_size=5
output_layer_size=3
```

```python
input_layer_size
```

```python
np.random.seed(42)
weights_input_hidden=np.random.uniform(size=(input_layer_size,hidden_layer_size))
weights_hidden_output=np.random.uniform(size=(hidden_layer_size, output_layer_size))
```

```python
print('weights_input_hidden',weights_input_hidden.shape,'weights_hidden_output',weights_hidden_output.shape)
```

```python
learning_rate=0.1
epochs=10000
```

```python
for epoch in range(epochs):
  input_layer=x_train
  hidden_layer_input=np.dot(input_layer, weights_input_hidden)
  hidden_layer_output=sigmoid(hidden_layer_input)
  output_layer_input=np.dot(hidden_layer_output,weights_hidden_output)
  predicted_output=sigmoid(output_layer_input)
  y_encoded=np.eye(output_layer_size)[y_train]
  error=y_encoded-predicted_output
  d_predicted_output=error*sigmoid_derivative(predicted_output)
  error_hidden_layer=d_predicted_output.dot(weights_hidden_output.T)
  d_hidden_layer=error_hidden_layer*sigmoid_derivative(hidden_layer_output)
  weights_hidden_output+=hidden_layer_output.T.dot(d_predicted_output)*learning_rate
  weights_input_hidden+=input_layer.T.dot(d_hidden_layer)*learning_rate
```

```python
hidden_layer_input=np.dot(x_test, weights_input_hidden)
hidden_layer_output=sigmoid(hidden_layer_input)
output_layer_input=np.dot(hidden_layer_output, weights_hidden_output)
predicted_output=sigmoid(output_layer_input)
```

```python
predicted_classes=np.argmax(predicted_output,axis=1)
predicted_classes
```

```python
accuracy=np.mean(predicted_classes==y_test)
print(f'accuracy={accuracy*100:.2f}%')
```
