```python
import numpy as np
import matplotlib.pyplot as plt
from keras.models import Sequential
from keras.layers import SimpleRNN,Dense
```

```python
def create_dataset(n_sample=200):
  x,y=[],[]
  for i in range(n_sample):
    seq=[j for j in range(i,i+3)]
    x.append(seq)
    y.append(i+3)
  return np.array(x),np.array(y)
```

```python
x,y = create_dataset(n_sample=200)
```

```python
x=x.reshape(x.shape[0],x.shape[1],1)
x
```

```python
model = Sequential()
model.add(SimpleRNN(20,activation='tanh' , input_shape = (x.shape[1] , 1)))
model.add(Dense(1))
```

```python
model.compile(optimizer='adam' , loss = 'mean_squared_error')
```

```python
history = model.fit(x,y,epochs=300,verbose=1)
```

```python
pred = model.predict(x)
```

```python
sample = np.array([[3], [4] , [5]]).reshape(1,3,1)
pred_val = model.predict(sample)
```

```python
print(f'predicted[3,4,5]:{pred_val[0][0] }')
print(f'actual=6')
```

```python



plt.plot(history.history['loss'] , label='loss')
plt.ylabel('loss')
plt.xlabel('epochs')
plt.legend()
plt.show()
```
