```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import  Sequential
from tensorflow.keras.layers import SimpleRNN,Dense,Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentence=['I love python', 'Deep learning is subset of machine learning',
          'I enjoy studing artifical intelligence',
          'Neual networks are fascinating',
          'Rnns are great for sequance prediction',
          'Rafiepour teach deep learning']
```

```python
max_words = 100
max_len=10
embeding_dim=8
```

```python
tokenizer = Tokenizer(num_words=max_words)
tokenizer.fit_on_texts(sentence)
sequences=tokenizer.texts_to_sequences(sentence)
```

```python
x=[]
y=[]
for seq in sequences:
  for i in range(1,len(seq)):
      x.append(seq[:i])
      y.append(seq[i])
```

```python
x=pad_sequences(x,maxlen=max_len)
y=np.array(y)
```

```python
model=Sequential()
model.add(Embedding(max_words,embeding_dim,input_length=max_len))
model.add(SimpleRNN(10,activation='tanh'))
model.add(Dense(max_words,activation='softmax'))
```

```python
model.compile(optimizer='adam' , loss = 'sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(x,y,epochs=400,verbose=1)
```

```python
test_sequance='Rafiepour teach'
test_seq=tokenizer.texts_to_sequences([test_sequance])[0]
test_seq=pad_sequences([test_seq],maxlen=max_len)
predicted=model.predict(test_seq)
predicted_word_index=np.argmax(predicted,axis=-1)[0]

```

```python
predicted_word=tokenizer.index_word[predicted_word_index]
print(f"predicted word='{test_sequance}'{predicted_word}")
```
