```python
import cv2
import numpy as np
from webcolors import rgb_to_name
import csv
```

```python


# تهیه دوربین
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # نمایش تصویر زنده
    cv2.imshow("Live Feed", frame)

    def get_color(event, x, y, flags, param):
        if event == cv2.EVENT_LBUTTONDOWN:
            color = frame[y, x]
            print(f"RGB: {color}")

    cv2.namedWindow("Live Feed")
    cv2.setMouseCallback("Live Feed", get_color)




    def rgb_to_color_name(rgb):
        try:
            return rgb_to_name(rgb)
        except ValueError:
            return "Unknown Color"

    

    with open("colors.csv", "w", newline='') as file:
        writer = csv.writer(file)
        writer.writerow(["Name", "RGB", "HEX"])
        writer.writerow(["Red", "(255, 0, 0)", "#FF0000"])
    
    # خروج با کلید 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break


       
cap.release()
cv2.destroyAllWindows()

```
